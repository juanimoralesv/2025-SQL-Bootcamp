````python
# Import libraries
import psycopg2 as pg2
````

````python
# Create connection to database
conn = pg2.connect(
    database = 'dvdrental',
    user = 'postgres',
    password = # Used and erased
)
````

A `cursor` is needed to traverse over the records of the database. It can be thought of as an pointer or an iterator.

````python
# Create the cursor object
cur = conn.cursor()
````

````python
# Query the database
cur.execute(
    "\
    SELECT * \
    FROM payment; \
    "
)
````

There are different methods for retrieving rows from the query:
- cur.fetchone() - to retrieve the first row
- cur.fetchmany() - to retrieve the first n rows, pass n
- cur.fetchall() - to retrieve all the rows

They return each row as a `list`, and inside each line there is a `tuple` with each of its elements being an cell in the row. In some cases, a `data type` is specified.

````python
# Retrieve ten rows
cur.fetchmany(10)
````

Once rows are fetched, it's possible to use tuple unpacking to access the data.

````python
# Store query results into a variable
data = cur.fetchmany(10)
data
````

````python
# Select a single cell/element from the table
data[2][4]
````

````python
# Close the connection
conn.close()
````

Avoid something like this:

```Python
add = 'DROP TABLE'
command = 'SELECT * FROM ' + add
cur.execute(command)
```

The variable `add` was going to be the name of a table, but for some reason it ended up having another text that could potentially damage the database.

According to best practices, it's better to write the query as a single line, or complete line. Not to build the query from parts of it.
