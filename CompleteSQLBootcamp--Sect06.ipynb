```python
# Función para tomar un archivo de consultas de SQL y formatear su texto
def query_text_format(path, file):
    # Cambio de directorio
    try:
        os.chdir(path)
    except FileNotFoundError:
        print(f"El directorio {path} no se encontró.")
        return
    
    # Lectura del archivo
    try:
        with open(file, 'r') as f:
            query_log = f.read()
    except FileNotFoundError:
        print(f"El archivo {file} no se encontró.")
        return
    
    # Formateo del texto
    # Remoción final de comentarios SQL
    query_log = query_log.replace(' */', '')
    # Separación en líneas
    query_log_lines = query_log.split('\n')
    
    # Iteración enumerada por todas las líneas del archivo
    for i, line in enumerate(query_log_lines):
        query_log_lines[i] = line.strip()
        
        # Remplazo de marcas de inicio de comentario SQL
        if query_log_lines[i][:2] == '/*':
            # Búsqueda de dos enteros para detectar comienzo de lección
            double_int = []            
            for char in line[3:5]:
                try:
                    if isinstance(int(char), int):
                        double_int.append(1)
                except:
                    pass
            
            # Si hay dos enteros, remplazo acorde a inicio de lección
            if len(double_int) == 2:
                query_log_lines[i] = query_log_lines[i].replace('/*', '###')
            # Si hay una letra, remplazo acorde a inicio de ejercicio en prueba
            elif query_log_lines[i][3].isalpha() and query_log_lines[i][4] == '.':
                query_log_lines[i] = query_log_lines[i].replace('/*', '####')
            # Demás comentarios de SQL son también comentarios de Python
            else:
                query_log_lines[i] = query_log_lines[i].replace('/*', '#')
        
        # Si la línea no está vacía
        elif line != '':
            query_log_lines[i] = '\t' + query_log_lines[i] + ' \\'
        
        # Demás casos (?)
        else:
            pass
        
    # Regresar el texto
    return '\n'.join(query_log_lines)
```

```python
# Función para tomar las consultas SQL y armar código en Python para ejecutarlas
def build_python_queries(text):
    # Separación en líneas
    query_lines = text.split('\n')
    
    # Armado de las consultas SQL en Python
    resulting_text = []
    flg_first_sql_line = True
    
    for i, line in enumerate(query_lines):
        # Línea vacía
        if line == '':
            resulting_text.append(line)
        # Línea que inicia con tabulación
        elif query_lines[i][0] == '\t':
            # Primera línea de consulta
            if flg_first_sql_line:
                # Primera línea de consulta de una única línea
                if (i + 1 <= len(query_lines) - 1) and ((query_lines[i + 1] == '') or ((query_lines[i + 1][0] != '') and (query_lines[i + 1][0] == '#'))):
                    resulting_text.append('query = " \\')
                    resulting_text.append(line)
                    resulting_text.append('\t"')
                    resulting_text.append('execute_query(query, conn)')
                # Primera línea de consulta de múltiples líneas
                else:
                    resulting_text.append('query = " \\')
                    resulting_text.append(line)
                    flg_first_sql_line = False
            # Segunda u otras líneas de consulta
            else:
                # Última línea de consulta
                if (i + 1 <= len(query_lines) - 1) and ((query_lines[i + 1] == '') or ((query_lines[i + 1][0] != '') and (query_lines[i + 1][0] == '#'))):
                    resulting_text.append(line)
                    resulting_text.append('\t"')
                    resulting_text.append('execute_query(query, conn)')
                    flg_first_sql_line = True
                # Línea de consulta y última absoluta
                elif i == len(query_lines) - 1:
                    resulting_text.append(line)
                    resulting_text.append('\t"')
                    resulting_text.append('execute_query(query, conn)')
                # Las líneas que no son ni primera ni última de consulta
                else:
                    resulting_text.append(line)
        # Otras líneas
        else:
            resulting_text.append(line)
    
    return '\n'.join(resulting_text)
```

```python
# Función para tomar el texto de una consulta SQL y ejecutarla
def execute_query(query_text, connection):
    # DataFrame desde consulta
    results_df = pd.read_sql_query(
        query_text,
        connection
    )
    # Impresión
    print(results_df)
```

```python



query = " \
	SELECT TO_CHAR(payment_date, 'Month-yyyy') \
	FROM payment; \
	"
execute_query(query, conn)

query = " \
	SELECT TO_CHAR(payment_date, 'Mon-yyyy') \
	FROM payment; \
	"
execute_query(query, conn)

query = " \
	SELECT TO_CHAR(payment_date, 'MONTH upalala! yyyy MI HH24') \
	FROM payment; \
	"
execute_query(query, conn)

query = " \
	SELECT TO_CHAR(payment_date, 'yyyy - Month - dd | HH24:MI') \
	FROM payment; \
	"
execute_query(query, conn)


### 50. Challenge: Timestamps and Extract
#### a. During which months did payments occur?
# Format your answer to return back the full month name.
query = " \
	SELECT TO_CHAR(payment_date, 'Month') AS payment_month \
	FROM payment \
	GROUP BY TO_CHAR(payment_date, 'Month'); \
	"
execute_query(query, conn)

# Or:
query = " \
	SELECT DISTINCT( \
	TO_CHAR(payment_date, 'Month') \
	) AS payment_month \
	FROM payment; \
	"
execute_query(query, conn)

#### b. How many payments ocurred on a monday?
query = " \
	SELECT TO_CHAR(payment_date, 'Day'), COUNT(payment_id) \
	FROM payment \
	GROUP BY TO_CHAR(payment_date, 'Day') \
	HAVING TO_CHAR(payment_date, 'Day') = 'Monday   '; \
	"
execute_query(query, conn)

# Or:
query = " \
	SELECT COUNT(*) \
	FROM payment \
	WHERE EXTRACT(DOW FROM payment_date) = 1; \
	"
execute_query(query, conn)


### 51. Mathematical Functions and Operators
# https://www.postgresql.org/docs/current/functions-math.html
# Explore the film table
query = " \
	SELECT * \
	FROM film \
	LIMIT 1; \
	"
execute_query(query, conn)

# Percentage of replacement_cost represented by rental_rate
query = " \
	SELECT ROUND((rental_rate / replacement_cost) * 100, 2) AS percent_cost \
	FROM film; \
	"
execute_query(query, conn)

# Deposit for replacement (10 %)
query = " \
	SELECT ROUND(0.1 * replacement_cost, 2) AS replacement_deposit \
	FROM film; \
	"
execute_query(query, conn)


### 52. String Functions and Operators
# https://www.postgresql.org/docs/17/functions-string.html
# Explore the customer table
query = " \
	SELECT * \
	FROM customer \
	LIMIT 1; \
	"
execute_query(query, conn)

# Length of first name
# LENGTH(), CHAR_LENGTH(), CHARACTER_LENGTH() are the same thing
query = " \
	SELECT LENGTH(first_name) AS name_length \
	"
execute_query(query, conn)

# Concatenation
query = " \
	SELECT first_name || ' ' || last_name AS full_name \
	FROM customer; \
	"
execute_query(query, conn)

# With upper casing
query = " \
	SELECT UPPER(first_name) || ' ' || UPPER(last_name) AS full_name \
	FROM customer; \
	"
execute_query(query, conn)

# Creation of email address
query = " \
	SELECT LEFT(LOWER(first_name), 1) || \
	'.' || \
	LOWER(last_name) || \
	'@dvdrental_customer.com' \
	AS customer_email \
	FROM customer; \
	"
execute_query(query, conn)

# Or:
query = " \
	SELECT LOWER( \
	LEFT(first_name, 1) || \
	'.' || \
	last_name || \
	'@dvdrental_customer.com' \
	) AS customer_email \
	FROM customer; \
	"
execute_query(query, conn)


### 53. Sub-Queries
# Explore the film table
query = " \
	SELECT * \
	FROM film \
	LIMIT 1; \
	"
execute_query(query, conn)

# Search the films with higher rental_rate than the average rental_rate
# (Subquery returns a single value)
query = " \
	SELECT title, rental_rate \
	FROM film \
	WHERE rental_rate > ( \
	SELECT AVG(rental_rate) \
	FROM film \
	); \
	"
execute_query(query, conn)

# Explore the rental table
query = " \
	SELECT * \
	FROM rental \
	LIMIT 1; \
	"
execute_query(query, conn)

# Explore the inventory table
query = " \
	SELECT * \
	FROM inventory \
	LIMIT 1; \
	"
execute_query(query, conn)

query = " \
	SELECT MAX(DATE(return_date)) \
	FROM rental; \
	"
execute_query(query, conn)

# Find films that have been returned between certain dates
query = " \
	SELECT title, return_date \
	FROM film \
	JOIN inventory \
	ON film.film_id = inventory.film_id \
	JOIN rental \
	ON inventory.inventory_id = rental.inventory_id \
	WHERE return_date BETWEEN '2005-05-29' \
	AND '2005-05-30' \
	ORDER BY return_date ASC; \
	"
execute_query(query, conn)

# Or:
query = " \
	SELECT film_id, title \
	FROM film \
	WHERE film_id IN ( \
	SELECT inventory.film_id \
	FROM inventory \
	JOIN rental \
	ON rental.inventory_id = inventory.inventory_id \
	WHERE return_date BETWEEN '2005-05-29' \
	AND '2005-05-30' \
	ORDER BY return_date ASC \
	); \
	"
execute_query(query, conn)
# Different results, because the first includes the possibility for a
# film to be returned more than once. The second one considers the unique
# values in the subquery of returned films.

# First query used as subquery and counting by title
query = " \
	SELECT title, COUNT(*) AS rental_count \
	FROM ( \
	SELECT title, return_date \
	FROM film \
	JOIN inventory ON film.film_id = inventory.film_id \
	JOIN rental ON inventory.inventory_id = rental.inventory_id \
	WHERE return_date BETWEEN '2005-05-29' AND '2005-05-30' \
	) \
	GROUP BY title \
	ORDER BY rental_count DESC; \
	"
execute_query(query, conn)
# There's a film that appears twice

# Second query used as subquery and counting by title
query = " \
	SELECT title, COUNT(*) AS rental_count \
	FROM ( \
	SELECT film_id, title \
	FROM film \
	WHERE film_id IN ( \
	SELECT inventory.film_id \
	FROM inventory \
	JOIN rental \
	ON rental.inventory_id = inventory.inventory_id \
	WHERE return_date BETWEEN '2005-05-29' \
	AND '2005-05-30' \
	ORDER BY return_date ASC \
	) \
	) \
	GROUP BY title \
	ORDER BY COUNT(*) DESC; \
	"
execute_query(query, conn)
# All films appear only once

# Find customers (first and last names) with at least
# one payment greater than 11
query = " \
	SELECT first_name, last_name \
	FROM customer AS c \
	WHERE EXISTS( \
	SELECT * \
	FROM payment AS p \
	WHERE p.customer_id = c.customer_id \
	AND amount > 11 \
	); \
	"
execute_query(query, conn)


### 54. Self-join
# Find all the pairs of films that have the same length
# Explore the film table
query = " \
	SELECT * \
	FROM film \
	LIMIT 1; \
	"
execute_query(query, conn)

# Query for the pairs
query = " \
	SELECT tleft.length, \
	tleft.title AS title_left, \
	tright.title AS title_right \
	FROM film AS tleft \
	JOIN film AS tright \
	ON tleft.length = tright.length \
	AND tleft.film_id != tright.film_id \
	ORDER BY tleft.length, tleft.title ASC; \
	"
execute_query(query, conn)
```
